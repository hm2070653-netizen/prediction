# Customer Churn Prediction

This notebook follows the project guide: explore telecom customer data, clean it, train baseline and stronger models, and evaluate churn predictions.

**Before running:** download the Telco Customer Churn CSV from Kaggle and place it in the `data/` folder in the project root.

## 1. Imports and project paths

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
DATA_DIR = ROOT / "data"
print("Project root:", ROOT)
print("Data folder exists:", DATA_DIR.exists())
```

## 2. Load the dataset

```python
csv_files = list(DATA_DIR.glob("*.csv"))
if not csv_files:
    print("No CSV found. Download the Telco Customer Churn dataset and place its CSV in data/.")
else:
    df = pd.read_csv(csv_files[0])
    display(df.head())
    print("Rows and columns:", df.shape)
```

## 3. Initial data inspection

```python
if "df" in globals():
    display(df.info())
    display(df.isna().sum().sort_values(ascending=False).head(15))
    if "Churn" in df.columns:
        display(df["Churn"].value_counts(dropna=False))
        display((df["Churn"].value_counts(normalize=True) * 100).round(2))
```

## 4. Explore churn patterns

```python
if "df" in globals() and "Churn" in df.columns:
    if "Contract" in df.columns:
        sns.countplot(data=df, x="Contract", hue="Churn")
        plt.title("Customer churn by contract type")
        plt.xticks(rotation=15)
        plt.tight_layout()
        plt.show()
    if "tenure" in df.columns:
        sns.histplot(data=df, x="tenure", hue="Churn", bins=25, element="step", stat="density", common_norm=False)
        plt.title("Tenure distribution by churn status")
        plt.tight_layout()
        plt.show()
    if "MonthlyCharges" in df.columns:
        sns.boxplot(data=df, x="Churn", y="MonthlyCharges")
        plt.title("Monthly charges by churn status")
        plt.tight_layout()
        plt.show()
```

## 5. Data cleaning preview

```python
if "df" in globals():
    clean_df = df.copy()
    if "TotalCharges" in clean_df.columns:
        clean_df["TotalCharges"] = pd.to_numeric(clean_df["TotalCharges"], errors="coerce")
    clean_df = clean_df.drop(columns=["customerID"], errors="ignore")
    if "Churn" in clean_df.columns:
        clean_df["Churn"] = clean_df["Churn"].map({"Yes": 1, "No": 0})
    display(clean_df.head())
    print("Missing values after TotalCharges conversion:")
    display(clean_df.isna().sum().sort_values(ascending=False).head(15))
```
